```python
from helper_chat import add_user_message, chat_response
```

```python
web_search_schema = {
    "type": "web_search_20250305",
    "name": "web_search",
    "max_uses": 5,
}
```

```python
def display_web_results(response):
    for block in response.content:
        if hasattr(block, 'type') and block.type == 'web_search_result':
            print(f"**{block.title}**")
            print(f"URL: {block.url}")
            if block.page_age:
                print(f"Page Age: {block.page_age}")
            print()
        elif hasattr(block, 'type') and block.type == 'text':
            print(block.text)
        elif hasattr(block, 'content') and isinstance(block.content, list):
            for item in block.content:
                if hasattr(item, 'type') and item.type == 'web_search_result':
                    print(f"**{item.title}**")
                    print(f"URL: {item.url}")
                    if item.page_age:
                        print(f"Page Age: {item.page_age}")
                    print()

```

```python
messages = []

add_user_message(
    messages,
    """
    Which AI model is market leader for image generation.
    """,
)

response = chat_response(messages, tools=[web_search_schema])

# Usage
display_web_results(response)

response
```
